```python
import pandas as pd
import numpy as np

print("PART 1")

# Загрузка данных Part 1
ranks_df = pd.read_csv("my_data/retrieval_ranks.csv")

def calculate_metrics(df, col):
    r1 = (df[col] == 1).mean()
    r3 = (df[col] <= 3).mean()
    r5 = (df[col] <= 5).mean()
    mrr = (1 / df[col].apply(lambda x: x if x <= 10 else np.inf)).mean()
    return r1, r3, r5, mrr

# 1a. Метрики для всех 60 вопросов
r1_A, r3_A, r5_A, mrr_A = calculate_metrics(ranks_df, "rank_A")
r1_B, r3_B, r5_B, mrr_B = calculate_metrics(ranks_df, "rank_B")

print("\n--- 1a. Metrics (all 60 questions) ---")
print(f"Config A -> recall@1: {r1_A:.3f}, recall@3: {r3_A:.3f}, recall@5: {r5_A:.3f}, MRR@10: {mrr_A:.3f}")
print(f"Config B -> recall@1: {r1_B:.3f}, recall@3: {r3_B:.3f}, recall@5: {r5_B:.3f}, MRR@10: {mrr_B:.3f}")

# 1b. Wilson Confidence Interval calculation
def wilson_interval(successes, total, z=1.96):
    p = successes / total
    denominator = 1 + (z**2) / total
    centre_adjusted = p + (z**2) / (2 * total)
    adjusted_std_dev = np.sqrt((p * (1 - p) / total) + (z**2) / (4 * total**2))
    lower = (centre_adjusted - z * adjusted_std_dev) / denominator
    upper = (centre_adjusted + z * adjusted_std_dev) / denominator
    return p, lower, upper

print("\n--- 1b. Recall and 95% Wilson Interval ---")
N = len(ranks_df)
for cfg, col in [("A, k=3", "rank_A"), ("A, k=5", "rank_A"), ("B, k=3", "rank_B"), ("B, k=5", "rank_B")]:
    k_val = int(cfg.split("k=")[1])
    succ = (ranks_df[col] <= k_val).sum()
    p, low, high = wilson_interval(succ, N)
    print(f"{cfg}: recall={p:.3f}, lower={low:.3f}, upper={high:.3f}")

# 1c. Annual model cost calculation
daily_requests = 11500
annual_requests = daily_requests * 365

# Трафик по языкам
shares = {'EN': 0.40, 'RU': 0.42, 'KK': 0.18}
sys_prompt = 318
q_tokens = {'EN': 33, 'RU': 53, 'KK': 66}
a_tokens = {'EN': 95, 'RU': 154, 'KK': 191}
chunk_A = {'EN': 241, 'RU': 390, 'KK': 484}
chunk_B = {'EN': 172, 'RU': 117, 'KK': 145}

p_input = 3.0 / 1e6
p_output = 18.0 / 1e6

def calc_annual_cost(chunk_dict, k):
    total_cost = 0
    for lang, share in shares.items():
        reqs = annual_requests * share
        inp_tok = reqs * (sys_prompt + q_tokens[lang] + k * chunk_dict[lang])
        out_tok = reqs * a_tokens[lang]
        total_cost += (inp_tok * p_input) + (out_tok * p_output)
    return round(total_cost)

print("\n--- 1c. Annual Model Cost (USD) ---")
print(f"A, k=3: ${calc_annual_cost(chunk_A, 3)}")
print(f"A, k=5: ${calc_annual_cost(chunk_A, 5)}")
print(f"B, k=3: ${calc_annual_cost(chunk_B, 3)}")
print(f"B, k=5: ${calc_annual_cost(chunk_B, 5)}")


print("\n=== PART 2 CALCULATIONS ===")

# Загрузка данных Part 2
docs_df = pd.read_csv("my_data/doc_results.csv")

# Точность по пайплайнам
acc_H = docs_df[docs_df['pipeline'] == 'H']['n_correct'].sum() / docs_df[docs_df['pipeline'] == 'H']['n_docs'].sum()
acc_L = docs_df[docs_df['pipeline'] == 'L']['n_correct'].sum() / docs_df[docs_df['pipeline'] == 'L']['n_docs'].sum()

# Слабые ячейки (weakest cell)
docs_df['acc'] = docs_df['n_correct'] / docs_df['n_docs']
weak_H = docs_df[docs_df['pipeline'] == 'H'].sort_values('acc').iloc[0]
weak_L = docs_df[docs_df['pipeline'] == 'L'].sort_values('acc').iloc[0]

p_h, low_h, high_h = wilson_interval(weak_H['n_correct'], weak_H['n_docs'])
p_l, low_l, high_l = wilson_interval(weak_L['n_correct'], weak_L['n_docs'])

# Стоимость Part 2
daily_pages = 9000
annual_pages = daily_pages * 365
personal_data_annual = round(annual_pages * 0.11)

# Pipeline H стоимость
img_in = 30 * 42  # 1260
img_out = 389
daily_cost_H = daily_pages * (img_in * (1.0 / 1e6) + img_out * (5.0 / 1e6))
annual_cost_H = round(daily_cost_H * 365)

# Pipeline L стоимость (3 сервера)
annual_cost_L = 3 * 750 * 12

print("\n--- 2a. Numbers ---")
print(f"Pipeline H -> Overall Accuracy: {acc_H:.3f}, Weakest: {weak_H['lang']} {weak_H['condition']} ({p_h:.3f} [{low_h:.3f}, {high_h:.3f}]), Annual Cost: ${annual_cost_H}")
print(f"Pipeline L -> Overall Accuracy: {acc_L:.3f}, Weakest: {weak_L['lang']} {weak_L['condition']} ({p_l:.3f} [{low_l:.3f}, {high_l:.3f}]), Annual Cost: ${annual_cost_L}")
print(f"Personal-data pages per year sent abroad under H: {personal_data_annual}")


print("\n=== PART 3 CALCULATIONS ===")

# Q1
cost_en = (1507 * 1.5 + 166 * 7.5) / 1e6
cost_kk = (3388 * 1.5 + 312 * 7.5) / 1e6
q1_bill_ratio = cost_kk / cost_en
print(f"Q1 Bill Ratio: {q1_bill_ratio:.2f}")

# Q2
q2_pos = round(896 * 0.65 * (263 / 256))
print(f"Q2 Position: {q2_pos}")

# Q3
logits = np.array([3.9, 3.0, 2.5, 1.6, 0.3])
probs = np.exp(logits) / np.sum(np.exp(logits))
print(f"Q3 Top Token Probability: {probs[0]*100:.1f}%")

# Q4
score_X = 68 * 1 + 32 * (-2) + 0 * 0
score_Y = 64 * 1 + 15 * (-2) + 21 * 0
q4_diff = score_Y - score_X
print(f"Q4 Y - X Penalty Score: {q4_diff}")

# Q5
q5_ratio = (11.62e12 / 70e9) / (1.4e12 / 70e9)
print(f"Q5 Tokens per parameter ratio: {q5_ratio:.2f}")
```
